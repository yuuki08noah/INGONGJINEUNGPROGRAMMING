#### **Lab 6. 행렬 곱은 '선형 변환(Linear Transformation)'이다**

행렬이 벡터 공간을 어떻게 왜곡(회전/확대)시키는지 시각화합니다.

```
import matplotlib.pyplot as plt

# 원본 벡터 (파란색 점)
v = np.array([1, 0])

# 변환 행렬 (반시계 90도 회전 행렬)

theta = np.radians(90)
R = np.array([[np.cos(theta),              ],
              [              ,  np.cos(theta)]])

# 선형 변환 수행 (행렬-벡터 곱)
v_rotated = np.dot(R, v)

# 시각화
plt.figure(figsize=(4,4))
plt.axhline(0, color='gray'); plt.axvline(0, color='gray')
plt.arrow(0, 0, v[0], v[1], head_width=0.1, color='blue', label='Original')
plt.arrow(0, 0, v_rotated[0], v_rotated[1], head_width=0.1, color='red', label='Transformed')
plt.xlim(-1.5, 1.5); plt.ylim(-1.5, 1.5)
plt.grid(); plt.legend()
plt.title("Linear Transformation (Rotation)")
plt.show()
```
